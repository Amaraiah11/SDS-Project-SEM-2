```python
gender_counts = mobiles['talk_time'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Time based on talktime')
plt.show()
```

```python
sns.scatterplot(x='battery_power', y='talk_time',hue='touch_screen',  data=mobiles, alpha=0.8)
plt.title('Battery POwer vs Talk time in smart phones')
plt.xlabel('battery_power')
plt.ylabel("talk_time")
plt.grid(True)
plt.show()
```

```python
# Scatter Plot: Price vs Battery Capacity
plt.figure(figsize=(7, 5))
sns.scatterplot(x=mobiles['battery_power'], y=mobiles['mobile_wt'], hue=mobiles['clock_speed'], palette='tab10')
plt.title("Battery Capacity vs mobile_wt")
plt.xlabel("Battery Capacity (mAh)")
plt.ylabel("Price in India")
plt.grid(True)
plt.show()
```

```python
# Scatter Plot: Screen Size vs Mobile Weight
plt.figure(figsize=(7, 5))
sns.scatterplot(x=mobiles['sc_h'], y=mobiles['mobile_wt'], hue=mobiles['talk_time'], palette='Set2')
plt.title("Screen Size vs Mobile Weight")
plt.xlabel("Screen Size (inches)")
plt.ylabel("Mobile Weight (g)")
plt.grid(True)
plt.show()
```
